```python
import os, random, shutil

src_img = '/kaggle/input/datasets/sahabipin/bangladeshi-native-vehicle-dataset/ODS_New/images'
src_lbl = '/kaggle/input/datasets/sahabipin/bangladeshi-native-vehicle-dataset/ODS_New/labels'

dst_root = '/kaggle/working/dataset'

# get matching image/label pairs (by filename stem)
images = sorted(os.listdir(src_img))
random.seed(42)
random.shuffle(images)

n = len(images)
train_end = int(n * 0.7)
valid_end = int(n * 0.85)  # 70/15/15 split

splits = {
    'train': images[:train_end],
    'valid': images[train_end:valid_end],
    'test':  images[valid_end:]
}

for split, files in splits.items():
    img_dir = os.path.join(dst_root, split, 'images')
    lbl_dir = os.path.join(dst_root, split, 'labels')
    os.makedirs(img_dir, exist_ok=True)
    os.makedirs(lbl_dir, exist_ok=True)

    for img_file in files:
        stem = os.path.splitext(img_file)[0]
        lbl_file = stem + '.txt'

        src_img_path = os.path.join(src_img, img_file)
        src_lbl_path = os.path.join(src_lbl, lbl_file)

        # only link if the matching label actually exists
        if os.path.exists(src_lbl_path):
            os.symlink(src_img_path, os.path.join(img_dir, img_file))
            os.symlink(src_lbl_path, os.path.join(lbl_dir, lbl_file))
        else:
            print('missing label for:', img_file)

    print(split, ':', len(files), 'images processed')
```

```python
for split in ['train', 'valid', 'test']:
    n_img = len(os.listdir(f'/kaggle/working/dataset/{split}/images'))
    n_lbl = len(os.listdir(f'/kaggle/working/dataset/{split}/labels'))
    print(split, 'images:', n_img, 'labels:', n_lbl)
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')  # nano model — fastest, good for pipeline testing

results = model.train(
    data='/kaggle/working/bnvd.yaml',
    epochs=5,          # just enough to confirm it works
    batch=16,
    imgsz=640,
    project='/kaggle/working/runs',
    name='bnvd_test'
)
```

```python
from ultralytics import YOLO

model = YOLO('yolov8m.pt')  # medium model, matches the paper's best-performing config

results = model.train(
    data='/kaggle/working/bnvd.yaml',
    epochs=100,
    batch=32,          # start here; drop to 16 if you hit CUDA out-of-memory
    imgsz=640,
    optimizer='SGD',
    lr0=0.01,
    lrf=0.01,
    momentum=0.937,
    weight_decay=0.0005,
    warmup_momentum=0.8,
    warmup_bias_lr=0.1,
    iou=0.7,
    patience=30,       # stops early if no improvement for 30 epochs, saves time
    save_period=10,     # saves a checkpoint every 10 epochs, not just at the end
    project='/kaggle/working/runs',
    name='bnvd_full'
)
```
